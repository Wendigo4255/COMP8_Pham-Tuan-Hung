```python
#w8a1
class Cylinder:
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height
        self.pi = 3.14
    def dttp (self):
        return 2 * self.pi * (self.radius + self.height)
    def volume (self):
        return self.pi * (self.radius ** 2) * self.height

r, h = map(int, input().split())
c = Cylinder(r, h)
print(f"{c.dttp():.2f}, {c.volume():.2f}")
```

```python
#w8a3
import math
class Point:
    def __init__ (self, x, y):
        self.x = x
        self.y = y
    
    def check_goc_toado (self):
        return self.x == 0 and self.y == 0
    
    def tung (self):
        return self.x == 0
    
    def hoanh (self):
        return self.y == 0
    
    def dist (self, other):
        return f"{math.sqrt((self.x - other.x) ** 2 + (self.y - other.y) ** 2):.2f}"

x, y = map(int, input().split())
p = Point(x, y)
if p.check_goc_toado():
    print("M la goc toa do")
elif p.tung():
    print("M nam tren truc tung")
elif p.hoanh():
    print("M nam tren truc hoanh")
else: print("M khong nam tren truc nao")

g = Point(0, 0)
print(p.dist(g))
```

```python
#w8a6
class rectangle:
    def __init__(self, cd, cr, k):
        self.cd = cd
        self.cr = cr
        self.k = k
    def scale(self):
        self.cr = self.k * self.cr
        self.cd = self.k * self.cd
    def area(self):
        return self.cr * self.cd
    def perimeter(self):
        return 2*(self.cr + self.cd)
a, b ,c = map(float, input().split())
n = rectangle(a, b, c)
n.scale()
x = n.area()
y = n.perimeter()
print("{:.2f}".format(x), "{:.2f}".format(y))
```

```python
#w8a7
import math
cd = list(map(str, input().split()))
def rutgon(x, y):
    k = math. gcd(x, y)
    x = x // k
    y = y //k
    return(x, y)
    
class fraction:
    def __init__(self, num, den, num2, den2):
        self.num, self.den = rutgon(num, den)
        self.num2 , self.den2 = rutgon(num2, den2)
        
    def add(self):
        mauchung = math.lcm(self.den,self.den2)
        tu = self.num * (mauchung // self.den) + self.num2 * (mauchung // self.den2)
        self.tu, self.mau = rutgon(tu, mauchung)
        return self
    
    def sub(self):
        mauchung = math.lcm(self.den,self.den2)
        tu = self.num * (mauchung // self.den) - self.num2 * (mauchung // self.den2)
        self.tu, self.mau = rutgon(tu, mauchung)
        return self
    
    def mul(self):
        tu = self.num * self.num2
        mau = self.den * self.den2
        self.tu, self.mau = rutgon(tu, mau)
        return self
    
    def div(self):
        tu = self.num * self.den2
        mau = self.num2 * self.den
        self.tu , self.mau  = rutgon(tu, mau)
        return self
    
    def __str__(self):
        if self.mau < 0:
            self.tu = -self.tu
            self.mau = -self.mau
        return f"{self.tu}/{self.mau}"
    
a = int(cd[0])
b = int(cd[1])
o = cd[2]
c = int(cd[3])
d = int(cd[4])
use = fraction(a,b,c,d)
if o == '+':
    print(use.add())
elif o == '-':
    print(use.sub())
elif o == '*':
    print(use.mul())
elif o == '/':
    print(use.div())
```

```python
#w8a8
class bankacc:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
    def deposit(self, money):
        self.balance += money
    def withdraw(self, money):
        if self.balance >= money:
            self.balance -= money
    def __str__(self):
        return f"{self.balance:.2f}"
name, ini_bal = map(str, input().split())
ini_bal = float(ini_bal)
inuse = bankacc(name, ini_bal)
n = int(input())
while n>0:
    x, money = map(str, input().split())
    money = float(money)
    if x == 'DEPOSIT':
        inuse.deposit(money)
    elif x == 'WITHDRAW':
        inuse.withdraw(money)
    n -= 1
print(inuse)
```

```python
#W8A9
class student:
    def __init__(self, name):
        self.name = name
        self.score = 0
        self.count = 0
    def add_score(self, a_score):
        self.score += a_score
        self.count += 1
        return self.score
    def avge(self):
        self.avg = self.score / self.count
        return f'{self.avg:.2f}'
    def rank(self):
        if self.avg >= 8:
            return "Excellent"
        elif 6.5 <= self.avg <8:
            return "Good"
        elif 5<= self.avg <6.5:
            return "Average"
        else:
            return "Poor"
name = input()
x = int(input())
inuse = student(name)
while x>0:
    a, b = map(str, input().split())
    b = float(b)
    inuse.add_score(b)
    x -= 1
a = inuse.avge()
b = inuse.rank()
print(name, a, b)
    
```

```python
#w8a10
class book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
class library:
    def __init__(self):
        self.lib = []
    def add(self, book):
        self.lib.append(book) 
    def count_author(self, x):
        count = 0
        for i in self.lib:
            if i.author == x:
                count += 1
        return count
    def count_year(self, y):
        count = 0
        for x in self.lib:
            if x.year == y:
                count += 1
        return count
x = int(input())
use = library()
while x > 0:
    s1, s2 = map(str, input().split())
    if s1 == "COUNT":
        print(use.count_author(s2))
    elif s1 == "COUNTYEAR":
        print(use.count_year(s2))
    else:
        input_book = list(s2.split(";"))
        input_book = book(input_book[0], input_book[1], input_book[2])
        use.add(input_book)
    x-=1

    
```

```python
#w8a11
import math
def plat_rastop(p1,p2,p3):
    (x1, y1, z1) = p1
    (x2, y2, z2) = p2
    (x3, y3, z3) = p3

    v1x, v1y, v1z = x2 - x1, y2 - y1, z2 - z1
    v2x, v2y, v2z = x3 - x1, y3 - y1, z3 - z1

    a = v1y * v2z - v1z * v2y
    b = v1z * v2x - v1x * v2z
    c = v1x * v2y - v1y * v2x
    
    h1 = (a, b, c)
    return h1
def plat_angle(q1,q2):
    (a,b,c) = q1
    (x,y,z) = q2
    dodai1 = math.sqrt(a**2+b**2+c**2)
    dodai2 = math.sqrt(x**2+y**2+z**2)
    vohuong = (a*x+b*y+c*z)
    h = math.acos(vohuong/(dodai1*dodai2))
    h = math.degrees(h)
    return f"{h:.2f}"
class point3:
    def __init__(self, A, B, C):
        self.A = A
        self.B = B
        self.C = C
    def toado(self):
        return (self.A, self.B, self.C)
    
x, y, z = map(float, input().split())
a = point3(x, y, z)
vect1 = a.toado()
x, y, z = map(float, input().split())
b = point3(x, y, z)
vect2 = b.toado()
x, y, z = map(float, input().split())
c = point3(x, y, z)
vect3 = c.toado()
x, y, z = map(float, input().split())
d = point3(x, y, z)
vect4 = d.toado()

rast1 = plat_rastop(vect1, vect2,vect3)
rast2 = plat_rastop(vect2, vect3,vect4)
print(plat_angle(rast1, rast2))
```

```python
#w8a13
import math
def triangle_check(a, b, c):
    if (a + b > c) and (a + c > b) and (b + c > a):
        return True
    return False
def square(a,b,c):
    p = (a + b + c) / 2
    k = math.sqrt(p * (p - a) * (p - b) * (p - c))
    return f"{k:.2f}"
class triangle:
    def __init__(self, edge1, edge2, edge3):
        self.edge1 = edge1
        self.edge2 = edge2
        self.edge3 = edge3
    def double_getarea(self):
        if triangle_check(self.edge1, self.edge2, self.edge3):
            return square(self.edge1, self.edge2, self.edge3)
        else:
            return "invalid"
x,y,z = map(int, input().split())
use= triangle(x,y,z)
print(use.double_getarea())
```
