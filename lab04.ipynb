```python
#W4a7
import math
def test_num(n):
  if n < 2:
    return False
  else:
    x = int(math.sqrt(n))
    for i in range(2, x +1):
      if n % i == 0:
        return False
  return True
def uoc_max(x):
  while True:
    if x <= 1:
      return 1
    for i in range(x + 1, 0, -1):
      if x % i == 0 and test_num(i):
        return i
    return 1
a = int(input())
print(f"Uoc nguyen to lon nhat cua so: {uoc_max(a)}")
```

```python
#w4a13
import math
def close_num(a, b):
  lst_1 = []
  lst_2 = []
  for i in range (1, a):
      if a % i == 0:
        lst_1.append(i)
  for i in range (1, b):
      if b % i == 0:
        lst_2.append(i)
  if a == sum(lst_2) and b == sum(lst_1):
      return True
  return False
(x, y) = map(int, input().split())
if close_num(x, y):
  print("true")
else:
  print("false")
```

```python
#w4a32
def chk_pass():
  n = input()
  if len(n) < 6:
    print("weak password")
    return
  chu_hoa = False
  chu_thg = False
  chu_so = False
  ki_tu = False
  for i in n:
    if '0' <= i <= '9':
      chu_so = True
    elif 'A' <= i <= 'Z':
      chu_hoa = True
    elif 'a' <= i <= 'z':
      chu_thg = True
    else:
      ki_tu = True
  if chu_thg and chu_hoa and chu_so and ki_tu:
    print("strong password")
  else:
    print("weak password")
chk_pass()
```

```python
#w4a33
def covrt_num():
  n = int(input())
  str_num = str(n)
  count = 0
  result = ""
  for i in range(len(str_num) - 1, -1, -1):
    result = str_num[i] + result
    count += 1
    if count % 3 == 0 and i != 0:
      result = "." + result
  print(result)
covrt_num()
```

```python
#w4a34
def xoa_chuoi():
    chuoi_a = input()
    chuoi_b = input()
    len_b = len(chuoi_b)
    if not chuoi_b:
        print(chuoi_a)
        return
    while True:
        vi_tri = chuoi_a.find(chuoi_b)
        if vi_tri == -1:
            break
        chuoi_a = chuoi_a[:vi_tri] + chuoi_a[vi_tri + len_b:]
    print(chuoi_a)
xoa_chuoi()
```
