```python
#w6a9
a = input().split()
b = input().split()
c= {}
for i in a:
  x, y = i.split(":")
  c[x] = int(y)
for i in b:
  p, q = i.split(":")

  if p in c:
    c[p] = c[p] + int(q)
  else:
    c[p] = int(q)
print(sorted(c))
```

```python
#w6a10
n = list(map(int, input().split()))
k = int(input())
seen_pair = []
result  = []
for i in range (len(n)):
  for j in range(i+1, len(n)):
    if n[i] + n[j] == k:
      pair = (n[i], n[j])
      if pair not in seen_pair:
        seen_pair.append(pair)
        result.append(pair)

result.sort()
print(result)
```

```python
#w6a12
n = list(map(int, input().split()))
num_dict = {}
for i in n:
  if i in num_dict:
    num_dict[i] += 1
  else:
    num_dict[i] = 1
max_req = 0
for count in num_dict.values():
  if count > max_req:
    max_req = count
same_req = []
for i, count in num_dict.items():
  if count == max_req:
    same_req.append(i)
result = min(same_req)
print(result)
```

```python
#w6a13
items = input()
tokens = items.split()
inverted_dict = {}

for i in range(0, len(tokens), 2):
    key_old = tokens[i]
    if i + 1 < len(tokens):
            val_old = tokens[i+1]
            inverted_dict[val_old] = key_old
print(inverted_dict)
```
