```python
#w9a1
def merge_lists(list1, list2):
    result = []
    i = 0 
    j = 0 
 
    while i < len(list1) and j < len(list2):
        if list1[i] < list2[j]:
            result.append(list1[i])
            i += 1
        else:
            result.append(list2[j])
            j += 1
            
    while i < len(list1):
        result.append(list1[i])
        i += 1
    while j < len(list2):
        result.append(list2[j])
        j += 1
        
    return result
l1 = list(map(int, input().split()))
l2 = list(map(int, input().split()))
print(f"{merge_lists(l1, l2)}")
```

```python
#w9a3
def minimize_number(n):
    s = list(str(n)) 
    length = len(s)
    
    for i in range(length):
        min_val = s[i]
        min_index = -1
        
        for j in range(i + 1, length):
            if s[j] < min_val: 
                min_val = s[j]
                min_index = j
            elif s[j] == min_val and min_index != -1:
                 min_index = j

        if min_index != -1:

            s[i], s[min_index] = s[min_index], s[i]
            break 
    return "".join(s)

A = int(input())
print(f"{minimize_number(A)}")
```

```python
#w9a4
def can_transform(start_str, end_str):
    i = 0 
    j = 0 
    while i < len(start_str):
  
        if j < len(end_str) and start_str[i] == end_str[j]:
            i += 1
            j += 1
        elif j < len(end_str) and start_str[i].upper() == end_str[j]:
            i += 1
            j += 1
        elif start_str[i].islower():
            i += 1
            return "NO"
    if j == len(end_str):
        return "YES"
    else:
        return "NO"
strA = input()
strB = input()
print(f"{can_transform(strA, strB)}")
```

```python
#w9a5
def is_sublist(list1, list2):
    n1 = len(list1)
    n2 = len(list2)

    for i in range(n2 - n1 + 1):
        sub_segment = list2[i : i + n1]
        if sub_segment == list1:
            return "YES"  
    return "NO"

l1 = list(map(int, input().split()))
l2 = list(map(int, input().split()))
print(f"{is_sublist(l1, l2)}")
```

```python
#w9a6
def next_permutation(nums):
    n = len(nums)
    i = n - 2

    while i >= 0 and nums[i] >= nums[i + 1]:
        i -= 1
    if i >= 0:
        j = n - 1
        while nums[j] <= nums[i]:
            j -= 1
        nums[i], nums[j] = nums[j], nums[i]
    left = i + 1
    right = n - 1
    while left < right:
        nums[left], nums[right] = nums[right], nums[left]
        left += 1
        right -= 1
    return nums

test = list(map(int, input().split()))
print(f"{next_permutation(test)}")
```

```python
#w9a7
def can_place_flowers(flowerbed, k):
    count = 0
    i = 0
    while i < len(flowerbed):
       
        if flowerbed[i] == 0:
            left_empty = (i == 0) or (flowerbed[i - 1] == 0)
            right_empty = (i == len(flowerbed) - 1) or (flowerbed[i + 1] == 0)
            if left_empty and right_empty:
                flowerbed[i] = 1 
                count += 1
                i += 1 
        i += 1
    return count >= k
lst_test = list(map(int, input().split()))
k_test = int(input())
print(f"{can_place_flowers(lst_test,k_test)}")
```

```python
#w9a8
def is_path_crossing(moves):
    x, y = 0, 0
    visited = set()
    visited.add((0, 0))
    
    for move in moves:
        if move == 'U': y += 1
        elif move == 'D': y -= 1
        elif move == 'R': x += 1
        elif move == 'L': x -= 1
        if (x, y) in visited:
            return True
        visited.add((x, y))
    return False
oper = input()
print(f"{is_path_crossing(oper)}")
```

```python
#w9a10
import bisect 
def find_radius(houses, heaters):
    houses.sort()
    heaters.sort()
    radius = 0
    
    for house in houses:
        
        i = bisect.bisect_left(heaters, house)
        left_dist = float('inf')
        if i > 0:
            left_dist = house - heaters[i - 1]

        right_dist = float('inf')
        if i < len(heaters):
            right_dist = heaters[i] - house
        min_dist = min(left_dist, right_dist)
        radius = max(radius, min_dist)
    return radius
test1 = list(map(int, input().split()))
test2 = list(map(int, input().split()))
print(f"{find_radius(test1, test2)}")
```
