```python
#w7a1
def binary_search():
    arr = list(map(int, input().split()))
    n = int(input())
    left  = 0
    right = len(arr)  -1
    while left <= right:
        mid = (left + right)//2
        if arr[mid] == n:
            return mid
        elif arr[mid] >n:
            right = mid -1
        else:
            left = mid + 1
    return -1
binary_search()
```

```python
#w7a3
def bubble_sort(arr):
  n = len(arr)
  for i in range(n):
    swap = False
    for j in range(0, n-i-1):
      if arr[j] > arr[j+1]:
        arr[j], arr[j+1] = arr[j+1], arr[j]
        swap = True
    if swap == False:
        break


def selction_sort(arr):
  n = len(arr)
  for i in range(n-1):
    min_idex = i
    for j in range(i+1, n):
      if arr[min_idex] > arr[j]:
        min_idex = j
    arr[i], arr[min_idex] = arr[min_idex], arr[i]


x = list(map(int, input().split()))
selction_sort(x)
print(x)
```

```python
#w7a8
def binary_search(arr, n):
    left  = 0
    right = len(arr)  -1
    while left <= right:
        mid = (left + right)//2
        if arr[mid][1] == n:
            return mid
        elif arr[mid][1] > n:
            right = mid -1
        else:
            left = mid + 1
    return -1
interval = eval(input())
interval.sort(key = lambda x: (x[1], x[1] - x[0] +1))
queries = eval(input())
for i in queries:
  print(binary_search(interval, i), end = " ")
```

```python
#w7a11
s1 = list(map(str, input().split()))
s2 = list(map(str, input().split()))

for i in range(len(s1)):
    while s1[i][-1].isalpha() == 0:
        s1[i] = s1[i][:-1]
cnt = 0
for i in range(0, len(s1) - len(s2) + 1):
    if s1[i:i + len(s2)] == s2:
        cnt += 1
print(cnt)
```
