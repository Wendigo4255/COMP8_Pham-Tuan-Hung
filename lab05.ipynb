```python
#w5a3
def prime_num():
  n = int(input())
  import math
  a = int(math.sqrt(n))
  if n < 2:
    return False
  for i in range (2, a +1):
    if n % i == 0:
      return False
      break
  return True
prime_num()
```

```python
#w5a4
def per_num():
  lst_num = []
  a = 0
  n = int(input())
  for i in range (1, n):
    if n % i == 0:
      lst_num.append(i)
  for x in lst_num:
    a += x
    if a == n:
      return True
  return False
per_num()
```

```python
#w5a5
def find_num():
  lst_num = list(map(int, input().split()))
  n = int(input())
  for i in range(len(lst_num)):
    if n == lst_num[i]:
      return i+1
  return -1
find_num()
```

```python
#w5a6
def fact():
  n = int(input())
  a = 1
  if n < 0:
    print("invalid")
  else:
    for i in range (1, n+1):
      a *= i
    return a
fact()
```

```python
#w5a7
def calculator():
  (num_1, operate, num_2) = input().split()
  result = 0
  if operate == '+':
    result = float(num_1) + float(num_2)
  if operate == '-':
    result = float(num_1) - float(num_2)
  if operate == '*':
    result = float(num_1) * float(num_2)
  if operate == '/':
    if float(num_2) == 0:
      return "invalid"
    result = float(num_1) / float(num_2)
  return "{:.2f}".format(result)
calculator()
```

```python
#w5a9
def sum_num():
  n = int(input())
  n = abs(n)
  a = 0
  while n > 0:
    b = n %10
    a += b
    n = n//10
  return a
sum_num()
```

```python
#w5a10
def kiem_tra_dang_cau(s1, s2):
    map_s1_to_s2 = {}
    map_s2_to_s1 = {}
    for c1, c2 in zip(s1, s2):

        if c1 in map_s1_to_s2:
            if map_s1_to_s2[c1] != c2:
                return False
        else:
            if c2 in map_s2_to_s1:
                return False

            map_s1_to_s2[c1] = c2
            map_s2_to_s1[c2] = c1

    return True
(x, y) = input().split()
kiem_tra_dang_cau(x, y)
```
