```python
#w10a1
def div(a, b):
    try:
        c = a/b
        return f'{c:.2f}'
    except ZeroDivisionError:
        return 'loi chia khong'
x, y = map(int,input().split())
print(div(x, y))
```

```python
#w10a3
def sq_sum(n):
    sum = 0
    while n != 0:
        i = n%10
        sum += i**2
        n = n // 10
    return sum


def happy_num(n):
    st = []
    while True:
        n = sq_sum(n)
        if n ==1:
            return True
        if n in st:
            return False
        st.append(n)

try:
    x = int(input())
    print(happy_num(x))
except ValueError:
    print("loi gia tri")
```

```python
#w10a6
def sqrt():
    import math
    try:
        n = int(input())
        a = math.sqrt(n)
        return a
    except ValueError:
        return 'so am'
print(sqrt())
```

```python
#w10a11
import re

def kiem_tra_regex():
    t = int(input())
    for _ in range(t):
            s = input()
            try:
                re.compile(s)
                print("True")
            except re.error:
                print("False")

kiem_tra_regex()
```
